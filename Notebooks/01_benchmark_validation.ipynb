# YorTIB-200 — Benchmark Validation

Checks the integrity and alignment of the Yoruba benchmark JSON and Excel versions.

```python
import json
from pathlib import Path
import pandas as pd

json_path = Path("../data/yortib_200.json")
xlsx_path = Path("../data/yortib_200.xlsx")

with open(json_path, encoding="utf-8") as f:
    data = json.load(f)
if isinstance(data, dict):
    data = data.get("data", data.get("items", []))

j = pd.DataFrame(data)
x = pd.read_excel(xlsx_path)

required = ["id","sentence","focus_word","options","answer"]
assert all(c in j.columns for c in required)
assert all(c in x.columns for c in required)

print("JSON:", j.shape)
print("Excel:", x.shape)
print("JSON duplicate IDs:", j["id"].duplicated().sum())
print("Excel duplicate IDs:", x["id"].duplicated().sum())
print("JSON IDs 1–200:", set(j.id) == set(range(1,201)))
print("Excel IDs 1–200:", set(x.id) == set(range(1,201)))
```

```python
for col in ["sentence","focus_word","answer"]:
    jj = j.set_index("id")[col].astype(str).sort_index()
    xx = x.set_index("id")[col].astype(str).sort_index()
    print(f"{col}: mismatches =", (jj != xx).sum())
```

```python
assert len(j) == 200
assert len(x) == 200
assert j.id.nunique() == 200
assert x.id.nunique() == 200
print("Benchmark integrity checks passed.")
```
