# YorTIB-200 — Gold Data Audit

Audits focus-word frequencies, semantic-label frequencies, repeated contexts, and benchmark composition.

```python
import json
from pathlib import Path
import pandas as pd
import numpy as np

DATA = Path("../data/yortib_200.json")
with open(DATA, encoding="utf-8") as f:
    data = json.load(f)
if isinstance(data, dict):
    data = data.get("data", data.get("items", []))
df = pd.DataFrame(data)
print("Rows:", len(df))
df.head()
```

```python
print("Focus-word distribution")
display(df["focus_word"].value_counts().rename_axis("focus_word").reset_index(name="N"))
```

```python
print("Semantic-label distribution")
display(df["answer"].value_counts().rename_axis("semantic_label").reset_index(name="N"))
```

```python
unique_sentence_focus = df[["sentence","focus_word"]].drop_duplicates()
print("Unique sentence + focus-word combinations:", len(unique_sentence_focus))
print("Repeated at sentence+focus level:", len(df) - len(unique_sentence_focus))

repeats = (
    df.groupby(["sentence","focus_word"])
      .size()
      .reset_index(name="N")
      .sort_values("N", ascending=False)
)
display(repeats.head(20))
```

```python
audit = pd.DataFrame({
    "metric": ["N", "unique IDs", "unique sentence+focus", "unique semantic labels"],
    "value": [len(df), df.id.nunique(), len(unique_sentence_focus), df.answer.nunique()]
})
display(audit)
```
